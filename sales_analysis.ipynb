```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../Dataset/SampleSuperstore.csv")

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

category_sales.plot(kind="bar", figsize=(6,4))
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

### Observation
- Technology generated the highest total sales.
- Office Supplies and Furniture also contributed significantly to total sales.

```python
region_profit = df.groupby("Region")["Profit"].sum()

region_profit.plot(kind="bar", figsize=(6,4))
plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")
plt.show()
```

### Observation
- Profit differed by region.
- Some regions generated higher profit, indicating stronger business performance.

```python
category_profit = df.groupby("Category")["Profit"].sum()

plt.figure(figsize=(8,5))
category_profit.plot(kind="bar", color=["red","blue","green"])

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.xticks(rotation=0)

plt.show()
```

### Observation
- Technology generated the highest profit.
- Furniture had lower profit compared to the other categories.

```python
import seaborn as sns

plt.figure(figsize=(8,5))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

### Observation

- The correlation heatmap shows the relationship between numerical variables in the dataset.
- Strong positive correlation indicates that two variables increase together.
- Weak correlation suggests that variables have little direct relationship with each other.
- The heatmap helps identify important relationships between Sales, Profit, Quantity, and Discount.

```python
ship_sales = df.groupby("Ship Mode")["Sales"].sum()

ship_sales.plot(kind="bar", figsize=(7,4))
plt.title("Sales by Ship Mode")
plt.xlabel("Ship Mode")
plt.ylabel("Sales")
plt.show()
```

### Observation
- Sales were distributed across different shipping modes.
- Certain shipping methods handled a larger share of total sales.

```python
region_sales = df.groupby("Region")["Sales"].sum()

plt.figure(figsize=(8,5))
region_sales.plot(kind="bar")

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.show()
```

### Observation
- The sales performance varies across different regions.
- Some regions generate significantly higher sales than others, highlighting regional differences in business performance.

```python
df["Sales"].plot(kind="hist", bins=30, figsize=(6,4))
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.show()
```

### Observation
- Most sales transactions were of lower value.
- A small number of high-value sales created a right-skewed distribution.

```python
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (10,6)
plt.rcParams['font.size'] = 12
```

```python
top10_states = df.groupby("State")["Sales"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
top10_states.plot(kind="bar", color="purple")

plt.title("Top 10 States by Sales")
plt.xlabel("State")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

plt.show()
```

### Observation

- The chart displays the top 10 states with the highest total sales.
- A few states contribute significantly more sales than others.
- These states represent the strongest-performing markets in the dataset.

```python
ship_profit = df.groupby("Ship Mode")["Profit"].sum()

plt.figure(figsize=(8,5))
ship_profit.plot(kind="bar", color="green")

plt.title("Total Profit by Ship Mode")
plt.xlabel("Ship Mode")
plt.ylabel("Total Profit")
plt.xticks(rotation=20)

plt.show()
```

### Observation

- Total profit varies across different shipping modes.
- Some shipping methods generate higher profit than others.
- This analysis helps identify the most profitable shipping options.
